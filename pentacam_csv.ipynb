```python
import os
import shutil

# ============================================================
# 1. DEFINE SOURCE AND DESTINATION FOLDERS
# ============================================================

source_folder = r"Z:\Pentacam 1 SSP HYD"

output_folder = r"C:\Pentacam 1 SSP HYD"


# ============================================================
# 2. CREATE OUTPUT MAIN FOLDER
# ============================================================

os.makedirs(output_folder, exist_ok=True)


# ============================================================
# 3. WALK THROUGH ALL SUBFOLDERS
# ============================================================

csv_count = 0

for root, dirs, files in os.walk(source_folder):

    # Find all CSV files in the current folder
    csv_files = [
        file for file in files
        if file.lower().endswith(".csv")
    ]

    # If there are no CSV files, continue
    if not csv_files:
        continue

    # --------------------------------------------------------
    # Get the relative path from the source folder
    # --------------------------------------------------------

    relative_path = os.path.relpath(root, source_folder)

    # Create the same folder structure in output
    destination_folder = os.path.join(
        output_folder,
        relative_path
    )

    os.makedirs(destination_folder, exist_ok=True)

    # --------------------------------------------------------
    # Copy each CSV file
    # --------------------------------------------------------

    for csv_file in csv_files:

        source_file = os.path.join(root, csv_file)
        destination_file = os.path.join(
            destination_folder,
            csv_file
        )

        shutil.copy2(source_file, destination_file)

        csv_count += 1

        print(f"Copied: {source_file}")
        print(f"     -> {destination_file}")


# ============================================================
# 4. SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("Extraction completed.")
print(f"Total CSV files copied: {csv_count}")
print(f"Output folder: {output_folder}")
print("=" * 60)
```
